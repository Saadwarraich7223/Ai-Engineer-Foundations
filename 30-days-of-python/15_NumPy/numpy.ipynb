## Lesson 1 - What is NumPy and Why It Exists

**What:** NumPy (Numerical Python) provides an N-dimensional array object and mathematical functions to operate on them at near-C speed.

**Why Python lists are not enough:**
- Lists store pointers to Python objects -> slow and memory-heavy
- `[1,2,3] * 2` duplicates the list, not the values
- No built-in vectorised math -> you must write loops

**Why NumPy:**
- Stores raw numeric data in contiguous memory blocks (like C arrays) -> ~50x faster
- Vectorised operations - operate on entire arrays without Python loops
- Foundation of: pandas, scikit-learn, TensorFlow, PyTorch, OpenCV, SciPy

**Where it is used:**

| Field | Use |
|-------|-----|
| Data Science | Feature matrices, cleaning, aggregation |
| Machine Learning | Training data, weight matrices |
| Deep Learning | Tensor math on CPU |
| Computer Vision | Images as arrays (H x W x C) |
| Signal Processing | FFT, filtering |
| Finance | Time-series, portfolio math |

```python
#Importing numpy
import numpy as np
import sys
import time

# Why NumPy? - Speed comparison
n=1_000_000

py_list=list(range(n))
t0=time.time()
py_result=[x*2 for x in py_list]
t_list=time.time()-t0

np_arr=np.arange(n)
t0=time.time()
np_result=np_arr*2
t_numpy=time.time()-t0

# Speed Comaprison
print(f"Python list  : {t_list*1000:.2f} ms")
print(f"NumPy array  : {t_numpy*1000:.2f} ms")
print(f"Speedup      : {t_list/t_numpy:.0f}x faster")

print(f"\nList memory  : {sys.getsizeof(py_list):,} bytes")
print(f"Array memory  : {sys.getsizeof(np_arr):,} bytes")
print(f"Memory ratio : {sys.getsizeof(py_list)/np_arr.nbytes:.1f}x smaller with NumPy")
```
