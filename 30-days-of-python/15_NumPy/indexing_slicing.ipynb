## Lesson 4 - Indexing and Slicing

**The most-used skill in daily NumPy work.** Mastering this lets you extract any data without unnecessary copies.

```python
import numpy as np
```

### 4.1 Basic Indexing - single elements

```python
a=np.arange(24).reshape(4,6)
print("Array:\n",a)

print('\nRow 0              :',a[0])
print('Row 2, col 4       :',a[2,4])
print("Las Row            :",a[-1])
print("Las element        :",a[-1,-1])
```

### 4.2 Slicing - ranges

```python
a=np.arange(24).reshape(4,6)

print("Rows 1-2:\n", a[1:3])
print("\nAll rows, cols 2-4:\n", a[:, 2:5])
print("\nEvery other col:\n", a[:, ::2])
print("\nReverse cols:\n", a[:, ::-1])

# Train/Test split pattern

X=np.random.randn(100,10).round(2)

X_train=X[:80]
X_test=X[80:]
print("\nTrain:", X_train.shape, " | Test:", X_test.shape)

```

### 4.3 Fancy (Integer) Indexing - arbitrary elements

```python
a=np.array([10,20,30,40,50,60])

# Select specific indices - used for batch sampling
indices=np.array([0,2,4])
print('Fancy Index:',a[indices])

# 2D: pair-wise selection
A=np.arange(20).reshape(4,5)
print(A)
rows=np.array([0,1,2])
cols=np.array([0,1,2])

print('Pair-wise:', A[rows,cols])

# Mini-batch sampling
dataset=np.random.randn(1000,20).round(2)

batch_idx=np.random.choice(1000,size=32,replace=False)
mini_batch=dataset[batch_idx]
print("Mini-batch shape:", mini_batch.shape)

```

### 4.4 Boolean (Mask) Indexing - filter by condition

```python
scores=np.array([72, 88, 45, 91, 67, 55, 95])
print('Passing(>=60)  :',scores[scores>=60])
print('60-90 range    :', scores[(scores>=60) & (scores <= 90)])

# Filter samples by class label
X=np.random.randn(100,4)
y=np.random.randint(0,3,size=100)
X_class0=X[y==0]
print("\nClass 0 samples:", X_class0.shape)

# Replace outliers - common in data cleaning
data=np.array([1.0, 200.0, 3.0, 4.0, 500.0])

# Left side ( what outlier to be replced) : right side (to replace with what )
data[data>100]=np.mean(data[data<=100])
print("After outlier fix:", data)

```

### 4.5 `np.where()` - Vectorised if-else
**When:** Conditional assignment, implementing activation functions, one-hot encoding.

```python
a = np.array([-3, -1, 0, 2, 5, -4, 7])

# ReLU activation - the most common use of np.where
result = np.where(a > 0, a, 0)
print("ReLU output:", result)

# Find indices where condition is True
idx = np.where(a < 0)
print("Negative indices:", idx[0])

# One hot encoding
label=3
n_classes=5
one_hot=np.where(np.arange(n_classes)==label,1,0)
print("One-hot:", one_hot)

```
