
## Lesson 6 - Math and Universal Functions (ufuncs)

**What:** Ufuncs operate element-wise on arrays at C speed.
**Why:** Replace Python loops entirely. Core of all vectorised ML math.

```python
import numpy as np
```

### 6.1 Element-wise arithmetic

```python
a = np.array([1., 2., 3., 4.])
b = np.array([10., 20., 30., 40.])

print("a + b :", a + b)
print("a - b :", a - b)
print("a * b :", a * b)      # element-wise, NOT matrix multiply
print("a / b :", a / b)
print("a ** 2:", a ** 2)
print("a % 3 :", a % 3)

# Scalar operations - broadcast automatically
print("\na * 5  :", a * 5)
print("a + 100:", a + 100)
```

### 6.2 Common mathematical ufuncs
**Includes activation functions used every day in ML.**

```python
x=np.linspace(0,2*np.pi,5)

print("sin :", np.sin(x).round(3))
print("sin :", np.cos(x).round(3))
print("sin :", np.exp(x).round(3))
print("log :", np.log(np.array([1, np.e, np.e**2])).round(3))
print("sqrt:", np.sqrt(np.array([1, 4, 9, 16])))
print("abs :", np.abs(np.array([-3, -1, 0, 2])))

z = np.array([-2., -1., 0., 1., 2.])

# Activation Functions to introduce non_linearity in deeplearning NN
# Sigmoid
sigmoid=1/(1+np.exp(z))
print("\nSigmoid:", sigmoid.round(3))

# ReLU
relu = np.maximum(0, z)
print("ReLU   :", relu)

# Softmax (with numerical stability trick)
def softmax(x):
    e = np.exp(x - x.max())   # subtract max to prevent overflow
    return e / e.sum()

logits = np.array([2.0, 1.0, 0.1])
print("Softmax:", softmax(logits).round(3), " | sum:", softmax(logits).sum())
```

### 6.3 Matrix multiplication - the core of ML

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Element-wise A*B:\n", A * B)      # NOT matrix multiply

# Matrix multiplication - 3 equivalent ways
print("\nA @ B:\n",          A @ B) 
print("np.dot(A,B):\n",    np.dot(A, B))
print("np.matmul(A,B):\n", np.matmul(A, B))

# Full forward pass of one neural layer
np.random.seed(0)
batch_size,n_in,n_out=32,128,64 

X=np.random.randn(batch_size,n_in)
W=np.random.randn(n_in,n_out) *0.01
print("length",(W))
b=np.zeros(n_out)

Z = X @ W + b                      # linear transformation
A_out = np.maximum(0, Z)   #Relue

print("\nNeural layer forward pass:")
print("  X:", X.shape, " W:", W.shape, " b:", b.shape)
print("  Z:", Z.shape, " A:", A_out.shape)
```
